```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob
```

```python
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (15, 10)

data_path = "./archive/dataset/dataset"

csv_files = glob.glob(f"{data_path}/*.csv")
print(f"Znaleziono {len(csv_files)} plików CSV:\n")
for i, file in enumerate(csv_files, 1):
    print(f"{i}. {os.path.basename(file)}")

print("\n" + "="*80)
print("ANALIZA KAŻDEGO PLIKU")
print("="*80)
```

```python
dataframes = {}

for csv_file in csv_files:
    filename = os.path.basename(csv_file)
    df = pd.read_csv(csv_file)
    dataframes[filename] = df
    
    print(f"\n📄 {filename}")
    print(f"   Wymiary: {df.shape[0]} wierszy × {df.shape[1]} kolumn")
    print(f"   Kolumny ({len(df.columns)}): {', '.join(df.columns[:5])}...")
    
    # Pokaż pierwsze 2 wiersze
    print(f"\n   Przykładowe dane:")
    print(df.head(2).to_string())
```

```python
print("\n" + "="*80)
print("SZCZEGÓŁOWA ANALIZA: Annual_P_L_1_final.csv")
print("="*80)

df = pd.read_csv(f"{data_path}/Annual_P_L_1_final.csv")

print(f"\n1️PODSTAWOWE INFORMACJE")
print(f"   Liczba firm: {df.shape[0]}")
print(f"   Liczba kolumn: {df.shape[1]}")
print(f"\n   Typy danych:")
print(df.dtypes.value_counts())

print(f"\n2️BRAKI DANYCH (top 10)")
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Brakujące': missing,
    'Procent': missing_percent
})
missing_df = missing_df[missing_df['Brakujące'] > 0].sort_values('Brakujące', ascending=False)
print(missing_df.head(10))

print(f"\n3️ROZKŁAD BRANŻ")
print(df['Industry'].value_counts().head(15))

print(f"\nSTATYSTYKI KLUCZOWYCH WSKAŹNIKÓW")
key_metrics = ['Current Price', 'Sales', 'OPM', 'Profit after tax', 
               'Return on capital employed', 'EPS']
print(df[key_metrics].describe())

print("\n" + "="*80)
print("POTENCJALNE PROBLEMY KLASYFIKACYJNE")
print("="*80)
```
